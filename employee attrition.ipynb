each department attrition category has been seperated

However, different dept. got different number of employees so in order to fix that we have to calculate the attrtion rate for each department.

```python
dept_attrition = pd.crosstab(
    df_cpy["Department"],
    df_cpy["Attrition"],
    normalize = "index")*100
print (dept_attrition)
#To visualize this result:
dept_attrition["Yes"].plot(kind="bar")
plt.title("Attrition rate by dept")
plt.xlabel("Department")
plt.ylabel("Employees who left (%)")
plt.show()
```

As we can see now the graph shows attrition rate according to depts. keeping in mind the number of people of each dept is different. Now we will check for the Overtime factor. Attrition rate for each overtime category:

```python

attrition_num = df_cpy["Attrition"].map(
    {"No": 0, "Yes": 1}
)

# attrition percentage for each overtime group
overtime_rate = (
    attrition_num.groupby(df_cpy["OverTime"]).mean() * 100
)

print(overtime_rate)
# To visualize this result 

overtime_rate.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")

plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# Making a copy of the cleaned dataset
df_model = df_cpy.copy()

# Converting Attrition into numerical values
df_model["Attrition"] = df_model["Attrition"].map(
    {"No": 0, "Yes": 1}
)

# Removing rows where the target value is missing or invalid
df_model = df_model.dropna(subset=["Attrition"])

# Separating the input features and target
X = df_model.drop(columns=["Attrition"])
y = df_model["Attrition"].astype(int)

# Removing the employee ID if it exists
if "EmployeeNumber" in X.columns:
    X = X.drop(columns=["EmployeeNumber"])

# Removing columns with only one distinct value
constant_columns = X.columns[X.nunique(dropna=False) <= 1]
X = X.drop(columns=constant_columns)

print("Input features:", X.shape) #X contains the input features that the model uses to make predictions, such as age, overtime and job satisfaction.
print("Target variable:", y.shape) #y contains the prediction of the model set and gives the output in 0 i.e for no and 1 for yes (attrition)


```

```python
# Identifying categorical columns
categorical_columns = X.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

# Identifying numerical columns
numerical_columns = X.select_dtypes(
    include=['number']
).columns.tolist()

print("Categorical columns:", categorical_columns)
print("\nNumerical columns:", numerical_columns)
```

```python
# Preprocessing for numerical columns
numerical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")) #fills missing values with the help of frequent median class
])

# Preprocessing for categorical columns
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")), #fills missing value with the help of most frequent data used for that column
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
]) #converts categorical values into binary columns. If an unfamiliar category appears during testing, it won't cause an encoding error.

# Applying the correct preprocessing to each column type
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)

print("Preprocessing is ready.")
```

# Training Two different models
                    1. With the help of logistic regression
                    2. With the help of random forest

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python

logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Training the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Creating the Random Forest model
random_forest_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced"
    ))
])

# Training the model
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
# Getting predictions from both models
logistic_predictions = logistic_model.predict(X_test)
forest_predictions = random_forest_model.predict(X_test)

# Function to calculate the evaluation metrics
def evaluate_model(actual, predicted, model_name):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(actual, predicted),
        "Precision": precision_score(
            actual, predicted, zero_division=0
        ),
        "Recall": recall_score(
            actual, predicted, zero_division=0
        ),
        "F1 Score": f1_score(
            actual, predicted, zero_division=0
        )
    }

# Comparing the results
results = pd.DataFrame([
    evaluate_model(y_test, logistic_predictions,
                   "Logistic Regression"),
    evaluate_model(y_test, forest_predictions,
                   "Random Forest")
])

results.round(3)
```

```python
# Creating the confusion matrix
cm = confusion_matrix(
    y_test,
    logistic_predictions,
    labels=[0, 1]
)

# Displaying it as a heatmap
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Stayed", "Left"],
    yticklabels=["Stayed", "Left"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
print(
    classification_report(
        y_test,
        logistic_predictions,
        labels=[0, 1],
        target_names=["Stayed", "Left"],
        zero_division=0
    )
)
```

```python
# Creating the confusion matrix for Random Forest
cm_rf = confusion_matrix(
    y_test,
    forest_predictions,
    labels=[0, 1]
)

# Displaying it as a heatmap
sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Stayed", "Left"],
    yticklabels=["Stayed", "Left"]
)

plt.title("Confusion Matrix - Random Forest")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
print(
    classification_report(
        y_test,
        forest_predictions,
        labels=[0, 1],
        target_names=["Stayed", "Left"],
        zero_division=0
    )
)
```

## Model Evaluation and Conclusion

In this stage, I developed two machine learning models for employee attrition prediction: Logistic Regression and Random Forest. Before training the models, I separated the input features from the target variable and divided the dataset into training and testing sets.

I used one-hot encoding to convert categorical data into numerical form and handled missing values wherever required. Both models were trained using the training dataset and evaluated using the testing dataset.

To compare the performance of the models, I used accuracy, precision, recall and F1-score. I also used a confusion matrix to understand the correct predictions and the mistakes made by the models.

Based on the evaluation results, I decided to select the model treained on random forest as it has comparatively more accuracy in the employees that stayed than logistic regression model
